```python
import json
from glob import glob
import os
types = {
    'gw': 'GW',
    'sa': 'SA',
    'qa': 'QA'
}
```

```python
if __name__ == "__main__":
    for t in types:
        for filename in glob(f'/mnt/c/Users/Beno/Downloads/{types[t]}/{types[t]}/*'):
            with open(filename, 'r') as frs:
                sol_data = json.load(frs)
            print(sol_data['instance'])
            problem_name = f'{sol_data['instance']}_J_log.json'
            with open(f'{t}/{problem_name}', 'r') as frp:
                problem_data = json.load(frp)
                
            problem_data['offset'] = round(problem_data['offset'])
            problem_data['solution'] = sol_data['x']
            problem_data['info'] = sol_data['additional_info']
            
            with open(f'{types[t]}/{problem_name}', 'w') as fw:
                json.dump(problem_data, fw)
        
        
        # print(filename)
```
